```python
import pandas as pd
import numpy as np
import os

df_clean = df.copy()

print("Original Shape:", df_clean.shape)
print("Missing Values:\n", df_clean.isnull().sum())
print("Duplicate Rows:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Cleaned Shape:", df_clean.shape)
print("Data Types:\n", df_clean.dtypes)
print("Statistics:\n", df_clean.describe())

save_path = r"C:\Users\AKASH-PC\Downloads\E-Commerce-Customer-Intelligence\Data\Processed"

os.makedirs(save_path, exist_ok=True)

df_clean.to_csv(
    os.path.join(save_path, "cleaned_customer_data.csv"),
    index=False
)

print("Cleaned Dataset Saved Successfully")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean["Purchase Date"] = pd.to_datetime(df_clean["Purchase Date"])

print("Total Customers:", df_clean["Customer ID"].nunique())
print("Total Revenue:", df_clean["Total Purchase Amount"].sum())
print("Total Orders:", len(df_clean))
print("Average Purchase:", df_clean["Total Purchase Amount"].mean())
print("Total Quantity:", df_clean["Quantity"].sum())
print("Churn Rate:", df_clean["Churn"].mean() * 100)

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.barplot(data=df_clean, x="Product Category", y="Total Purchase Amount", estimator=sum, ax=axes[0, 0])
axes[0, 0].set_title("Revenue by Product Category")
axes[0, 0].tick_params(axis="x", rotation=30)

sns.countplot(data=df_clean, x="Payment Method", ax=axes[0, 1])
axes[0, 1].set_title("Payment Method Distribution")

sns.histplot(data=df_clean, x="Customer Age", kde=True, ax=axes[1, 0])
axes[1, 0].set_title("Customer Age Distribution")

sns.countplot(data=df_clean, x="Churn", ax=axes[1, 1])
axes[1, 1].set_title("Customer Churn Distribution")

plt.tight_layout()
plt.show()

monthly_revenue = df_clean.groupby(
    df_clean["Purchase Date"].dt.to_period("M")
)["Total Purchase Amount"].sum()

monthly_revenue.plot(figsize=(12, 5), marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

snapshot_date = df_clean["Purchase Date"].max() + pd.Timedelta(days=1)

rfm = df_clean.groupby("Customer ID").agg(
    Recency=("Purchase Date", lambda x: (snapshot_date - x.max()).days),
    Frequency=("Purchase Date", "count"),
    Monetary=("Total Purchase Amount", "sum")
)

rfm["R_Score"] = pd.qcut(rfm["Recency"].rank(method="first"), 4, labels=[4, 3, 2, 1])
rfm["F_Score"] = pd.qcut(rfm["Frequency"].rank(method="first"), 4, labels=[1, 2, 3, 4])
rfm["M_Score"] = pd.qcut(rfm["Monetary"].rank(method="first"), 4, labels=[1, 2, 3, 4])

rfm["RFM_Score"] = rfm["R_Score"].astype(str) + rfm["F_Score"].astype(str) + rfm["M_Score"].astype(str)

rfm["Segment"] = "Regular Customers"
rfm.loc[rfm["RFM_Score"].isin(["444", "443", "434", "344"]), "Segment"] = "Champions"
rfm.loc[rfm["R_Score"].astype(int) <= 2, "Segment"] = "At Risk"
rfm.loc[rfm["R_Score"].astype(int) == 1, "Segment"] = "Lost Customers"

print(rfm.head())

sns.countplot(data=rfm, x="Segment")
plt.title("Customer Segmentation")
plt.xticks(rotation=30)
plt.show()

rfm.to_csv(
    r"C:\Users\AKASH-PC\Downloads\E-Commerce-Customer-Intelligence\Data\Processed\customer_segmentation.csv"
)
```
